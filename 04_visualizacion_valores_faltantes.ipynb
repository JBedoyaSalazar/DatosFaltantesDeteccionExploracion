# Visualizacion De Valores Faltantes

```python
import pandas as pd
import missingno as msno
import matplotlib.pyplot as plt

from src.pandas_accessors import MissingMethods
from src.data_loader import load_all_datasets

datasets = load_all_datasets()

riskfactors = datasets['riskfactors']
```

## Variable

```python
riskfactors.missing.missing_variable_plot()
```

## Casos / Observaciones / Filas

```python
riskfactors.missing.missing_case_plot()
```

```python
riskfactors.missing.missing_variable_span_plot(variable="weight_lbs", span_every=10, rot = 0)
```

```python
msno.bar(
    riskfactors,
    figsize=(14, 5),
    fontsize=9,
    color="#2b5c8f",
    sort="descending",
)

plt.title("Resumen de valores presentes por variable", fontsize=12, pad=20)
plt.show()
```

```python
msno.matrix(riskfactors,)

plt.title("Resumen de valores presentes por variable", fontsize=12, pad=20)
plt.show()
```

```python
riskfactors.missing.missing_heatmap_plot(
    variables=["pregnant", "bmi"]
)
```
